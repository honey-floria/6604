```python
# Perceptron demo
import numpy as np
import matplotlib.pyplot as plt
import copy
```

```python
# Part 1: Basic Demo of Perceptron
# Perceptron parameters
W12=-3
W13= 2
theta = -1  # the bias term
learnrate = 0.3

# Eight training instances are defined
tmpp= np.array([
    [1.0000  ,  1.0000   ,  1.0000],
    [1.0000  ,  0.5000  ,   1.0000],
    [1.5000  ,  0.2000  ,   1.0000],
    [0.5000  ,  0.7500  ,   1.0000],
    [0.5000  ,  0.3000  ,   0.0000],
    [0.5000  ,  -0.7000 ,   0.0000],
    [0.0000  ,  0.2    ,    0.0000],
    [-0.5    ,  0.5     ,   0.0000]
   ])
print(tmpp.shape)
```

```python
X2=tmpp[:,0]
X3=tmpp[:,1]
desiredout=tmpp[:,2]
print('X2 = ', X2)
print('X3 = ', X3)
print('desiredout = ',desiredout)
```

```python
# plot
Num_datapoints = len(X2)
plt.scatter(X2, X3, c=desiredout, cmap='viridis', edgecolors='k', s=30)
plt.title(f'Display of data points\n'
              f'Number of points: {Num_datapoints}')
plt.xlabel('Feature X2')
plt.ylabel('Feature X3')
plt.xlim(-1, 2)  # xmin=-1, xmax=2
plt.ylim(-1, 2)  
plt.show()
```

```python
# training algorithm for perceptron (2-dimentional case)
out = -1
for iteration in range(10):  # i=0,1,2,3,...,9
    totalerror=0
    for j in range(8):  # j=0,1,2,3,...7
        tmpout = W12*X2[j]+W13*X3[j] + theta
        if (tmpout > 0):
            out = 1
        else:
            out = 0
        diff = desiredout[j] - out       # calculate the error
        W12 = W12 + learnrate*diff*X2[j]; # error correction learning rule
        W13 = W13 + learnrate*diff*X3[j];
        theta = theta + learnrate*diff;
        totalerror = totalerror + np.abs(diff);
        #debug print(iteration, j, totalerror, W12, W13, theta)
        
```

```python
# Create meshgrid for decision boundary visualization
xx, yy = np.meshgrid(np.linspace(-1, 2, 100),
                     np.linspace(-1, 2, 100))
print('xx.shape =  ', xx.shape)
```

```python
# Get output for the grid points to plot decision boundaries
percept = copy.deepcopy(xx)

for ixx in range(100):  
    for iyy in range(100):  
        tmpout=W12*xx[ixx,iyy]+W13*yy[ixx,iyy]+theta
        if (tmpout > 0):
            percept[ixx,iyy] = 1
        else:
            percept[ixx,iyy] = 0
```

```python
# Visualize the decision boundary
threshold_1 = 0.5

plt.contourf(xx, yy, percept, levels=[0, threshold_1, 1], alpha=0.3, cmap=plt.cm.Greys)
plt.scatter(X2, X3, c=desiredout, cmap='viridis', edgecolors='k', s=30)

plt.title(f'Decision Boundary with threshold {threshold_1}\n')
plt.xlabel('Feature X2')
plt.ylabel('Feature X3')

plt.show()
```

```python
# Initialize the vectors to save the values
save_W12 = [-3]
save_W13 = [2]
save_theta = [-1]
save_totalerror = [0]

# Perceptron parameters
W12=-3
W13= 2
theta = -1  # the bias term
learnrate = 0.3

# training algorithm for perceptron (2-dimentional case)
out = -1
for iteration in range(10):  # i=0,1,2,3,...,9
    totalerror=0
    for j in range(8):  # j=0,1,2,3,...7
        tmpout = W12*X2[j]+W13*X3[j] + theta
        if (tmpout > 0):
            out = 1
        else:
            out = 0
        diff = desiredout[j] - out       # calculate the error
        W12 = W12 + learnrate*diff*X2[j]; # error correction learning rule
        W13 = W13 + learnrate*diff*X3[j];
        theta = theta + learnrate*diff;
        totalerror = totalerror + np.abs(diff);
        
        save_W12.append(W12)
        save_W13.append(W13)
        save_theta.append(theta)
        save_totalerror.append(totalerror)


# Plotting
print('Length of saved values is ',len(save_W12))
xaxis = list(range(len(save_W12)))


plt.plot(xaxis,save_W13,label = 'W13', marker = 'o')
plt.plot(xaxis,save_W12,label = 'W12', marker = 's')
plt.plot(xaxis,save_theta,label = 'theta', marker = '^')
plt.plot(xaxis,save_totalerror,label = 'totalerror', marker = '*')

#Add titles and labels
plt.title('Convergence of the parameters')
plt.xlabel('iteration number')
plt.ylabel('Values of W13,W12,b,totalerror')

plt.legend() # To show the legend for the curves
plt.grid()   # To show the grid
plt.ylim(-4, 8) 
plt.show()   # To display the plot
```
